```python
# 入力
H = 4
W = 4
S = [
    '#.#.'
    ,'....'
    ,'..#.'
    ,'....'
]
S
```

```python
S_list = []
for i in S:
    S_list.append((list(i)))
S_list
```

```python
init = [[0]*W for i in range(H)]
init
```

```python
'#' in S_list[0]
```

```python
[row[2] for row in S_list]
```

```python
for i in range(H):
    if '#' in S_list[i]:
        init[i] = [1]*W
init
```

```python
for i in range(W):
    if '#' in [row[i] for row in S_list]:
        for row in init:
            row[i] = 1
init
```

```python
sum(sum(row) for row in init)
```
