```python
# 入力
import numpy as np

N = 2
M = 3
base = np.array([400, 410, 420])
h = np.array([[400, 400, 400],[300, 300, 300]])
```

```python
# 正解からの差を計算
dif = h - base
dif
```

```python
abs(dif[0,1])
```

```python
deduction = []
deduction.append(0)
deduction.append(1)
deduction
```

```python
deduction = np.zeros_like(dif)
deduction
```

```python
# 減点計算
deduction = np.zeros_like(dif)
for n in range(N):
    for m in range(M):
        if abs(dif[n,m]) <= 5:
            deduction[n,m] += 0
        elif abs(dif[n,m]) <= 10:
            deduction[n,m] += 1
        elif abs(dif[n,m]) <= 20:
            deduction[n,m] += 2
        elif abs(dif[n,m]) <= 30:
            deduction[n,m] += 3
        else:
            deduction[n,m] += 5

deduction

deduction_person = np.sum(deduction, axis=1)
deduction_person
```

```python
# 得点計算
point = 100 - deduction_person
point
```

```python
# 最高点出力
print(np.maximum(0,point).max())
```
