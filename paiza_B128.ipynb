```python
# 入力
N = 123456
```

```python
# N個のリストに分割
N_list = list(map(int, str(N)))
N_list
```

```python
# 全体の枠設計
import numpy as np
n = len(N_list)
```

```python
sub_array = np.zeros((3,3), dtype=object)
sub_array[0] = ["#"]*3
sub_array
```

```python
# "#"入れる関数定義
def fill_num(num):
    sub_array = np.zeros((3,3), dtype=object)
    sub_array[sub_array == 0] = '.'
    a = num // 3
    b = num % 3
    if a == 0:
        for i in range(b):
            sub_array[0,i] = "#"
    elif a == 1:
        sub_array[0] = ["#"]*3
        for i in range(b):
            sub_array[1,i] = "#"
    else:
        sub_array[0] = ["#"]*3
        sub_array[1] = ["#"]*3
        for i in range(b):    
            sub_array[2,i] = "#"
    
    return sub_array.tolist()
```

```python
fill_num(5)
```

```python
a = []
for i in N_list:
    a.append(fill_num(i))

a
```

```python
array_a = np.array(a)
array_a
```
