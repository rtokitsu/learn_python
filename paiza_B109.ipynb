```python
# 入力
N = 4
H = 3
W = 2
P = 2
Q = 0

occupied_seat = {(0,0), (1,0), (1,1), (2,1)}
```

```python
# 全席の座標を作成
all_seat = set()
for i in range(H):
    for j in range(W):
        all_seat.add((i,j))

all_seat
```

```python
# 予約されてる席を除外して、予約されてない席の座標を取得
available_seat = all_seat - occupied_seat
available_seat
```

```python
# それぞれのマンハッタン距離を計算
distance = {seat: abs(seat[0] - P) + abs(seat[1] - Q) for seat in available_seat}
distance    
```

```python
# マンハッタン距離が最小のやつの座標を取得
min_distance = min(distance.values())
output = [(h, w) for h, w in distance.items() if w == min_distance]
output_final = set(dict(output).keys())
output_final
```

```python
# 出力
for x,y in output_final:
    print(x,y)
```
