```python
# inputのN x Nの２次元リストを作成（list_1）
N = 5
list_1 = [[3,3,3,2,6],
          [1,9,4,5,3],
          [1,8,4,6,5],
          [4,9,2,6,8],
          [7,5,8,1,6]
          ]
```

```python
# 最終形態（ピラミッド）のN x Nの２次元リストを作成（list_2）
list_2 = []
for i in range(1, N+1):
    if i <= N/2:
        lift_2_sub = []
        for j in range(1, N+1):
            if i <= j <= N - i + 1:
                lift_2_sub.append(i)
            elif j <= N/2:
                lift_2_sub.append(j)
            else:
                lift_2_sub.append(N - j + 1)
        list_2.append(lift_2_sub)
    else:
        lift_2_sub = []
        for j in range(1, N+1):
            if N - i + 1<= j <= i:
                lift_2_sub.append(N - i + 1)
            elif j <= N/2:
                lift_2_sub.append(j)
            else:
                lift_2_sub.append(N - j + 1)
        list_2.append(lift_2_sub)

list_2
```

```python
# list_1とlist_2の要素ごとの差分を取る（list_3）
import numpy as np
list1_np = np.array(list_1)
list2_np = np.array(list_2)
list3_np = list1_np - list2_np
list3_np
```

```python
# list_3の各要素の合計を計算
print(list3_np.sum())
```
