```python
# 入力
N = 6
cars = [4,1,5,6,2,3]
```

```python
count = 0
while cars:
    eval = []
    for i in range(len(cars)):
        if cars[i] == min(cars[i:len(cars)]):
            eval.append(True)
        else:
            eval.append(False)
    cars = [c for c, f in zip(cars, eval) if not f] 
    count += 1
print(count-1)
```
