```python
count = 0
while cars:
    eval = []
    for i in range(len(cars)):
        if cars[i] == min(cars[i:len(cars)]):
            eval.append(True)
        else:
            eval.append(False)
    cars = [c for c, f in zip(cars, eval) if not f] 
    count += 1
print(count-1)
```

```python
def count_rounds(N, A):
    # 各車の位置を記録
    position = [0] * (N + 1)
    for i in range(N):
        position[A[i]] = i  # 車番 A[i] の位置を記録

    # 周回数のカウント
    rounds = 0
    current_position = position[1]  # 1番の車の位置

    for num in range(2, N + 1):
        if position[num] < current_position:
            rounds += 1  # 逆走するので周回が必要
        current_position = position[num]  # 現在位置を更新

    print(rounds)

# 入力を受け取る
# N = int(input().strip())
# A = [int(input().strip()) for _ in range(N)]

# 関数を呼び出して結果を出力
count_rounds(6, [6,5,4,1,2,3])
```
