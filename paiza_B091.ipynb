```python
# 入力
N = 3
H = [
    [90,10,10]
    ,[10,30,20]
    ,[10,10,20]
]
```

```python
output = []
for i in range(N):
    for j in range(N):
        if i == 0:
            if j == 0:
                if H[i][j] > H[i][j+1]:
                    if H[i][j] > H[i+1][j]:
                        output.append(H[i][j])
            elif j == N-1:
                if H[i][j] > H[i][j-1]:
                    if H[i][j] > H[i+1][j]:
                        output.append(H[i][j])
            else:
                if H[i][j] > H[i][j-1]:
                    if H[i][j] > H[i][j+1]:
                        if H[i][j] > H[i+1][j]:
                            output.append(H[i][j])
        elif i == N-1:
            if j == 0:
                if H[i][j] > H[i][j+1]:
                    if H[i][j] > H[i-1][j]:
                        output.append(H[i][j])
            elif j == N-1:
                if H[i][j] > H[i][j-1]:
                    if H[i][j] > H[i-1][j]:
                        output.append(H[i][j])
            else:
                if H[i][j] > H[i][j-1]:
                    if H[i][j] > H[i][j+1]:
                        if H[i][j] > H[i-1][j]:
                            output.append(H[i][j])
        else:
            if j == 0:
                if H[i][j] > H[i][j+1]:
                    if H[i][j] > H[i-1][j]:
                        if H[i][j] > H[i+1][j]:
                            output.append(H[i][j])
            elif j == N-1:
                if H[i][j] > H[i][j-1]:
                    if H[i][j] > H[i-1][j]:
                        if H[i][j] > H[i+1][j]:
                            output.append(H[i][j])
            else:
                if H[i][j] > H[i][j-1]:
                    if H[i][j] > H[i][j+1]:
                        if H[i][j] > H[i-1][j]:
                            if H[i][j] > H[i+1][j]:
                                output.append(H[i][j])
output
```
